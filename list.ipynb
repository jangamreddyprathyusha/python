max of list

```python
l=[10,20,60,30,40]
max=0
for val in l:
    if max<val:
      max=val
print(max)
```

min of list

```python
l=[10,20,60,30,40]
min=l[0]
for val in l:
    if min>val:
      min=val
print(min)
```
