All the athmetic operators

```python
add=lambda a,b:a+b
print(add(10,20))
```

```python
sub=lambda a,b:a-b
print(sub(12,9))
```

```python
mul=lambda a,b:a*b
print(mul(19,12))
```

```python
div=lambda a,b:a/b
print(div(10,20))
```

```python
fdiv=lambda a,b:a//b
print(fdiv(10,6))
```

```python
mod=lambda a,b:a%b
print(mod(20,6))
```

position arguments

```python
add=lambda a,b:a+b
print(add(a=12,b=30))
```

```python
add=lambda a,b,c:a+b+c
print(add(13,c=20,b=30))
```

keyword arguments

```python
sub=lambda a,b:a-b
print(sub(b=20,a=10))
```

```python
add=lambda a,b,c:a*b+c
print(add(13,c=20,b=30))
```

default arguments

```python
add=lambda a,b=12:a+b
print(add(9))
```

```python
mul=lambda a=14,b=12:a-b
print(mul())
```

```python
def my_map(func,it):
    res=[]
    for val in it:
       res.append(func(val))
    return res
def cube(val):
    return val**3
l=[10,20,30,40]
result=my_map(cube,l)
print(result)
    
```

```python
def Toupper(val):
    st=""
    for ch in val:
      st+=chr(ord(ch)-32)
    return st
def my_map(func,string):
    res=[]
    for val in string:
       res.append(func(val))
    return res
st=["priya","sneha","pandu","lara"]
upper=my_map(Toupper,st)
print(upper)
```

```python
num=[10,20,30,40,50,60]
f=filter(lambda x:x%3==0,num)
print(list(f))
```

```python
def divby3(val):
    if(val%3==0):
        return True
    else:
        return False
def my_map(func,it):
    res=[]
    for val in it:
        if(func(val)==True):
            res.append(val) 
    return res
num=[10,20,30,40,50,60]
result=my_map(divby3,num)
print(result)   
```
