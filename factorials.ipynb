Recursion :recursion function calls itself
1.base case-->stops the recursion
2.recursive-->calls itself

```python
n=5
def factorial(n):
    if n==1:
       return 1
    else:
       return n*factorial(n-1)
factorial(n)
```

6.using while

```python
n=5
fact=1
i=1
while i<=n:
    fact=fact*i
    i+=1
print(fact)
```

7.check whether a number is a factorial

```python
n=120
fact=1
i=1
while fact<n:
    fact=fact*i
    i+=1
if fact==n:
    print("factorial number")
else:
    print("not a factorial")
```
