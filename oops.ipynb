Instance methods with attributes

```python
class student:
    def __init__(self,name,age):  #self--->referce to the object
        self.name=name
        self.age=age
    def display(self):   #instance methods:works with a particular object.
        print(self.name)
        print(self.age)
s1=student("prathu",21)   #works with different objects.   --->thats why it is called an instance method.
s2=student("reena",22)
s1.display()
s2.display()
```

class method:works with the class itself,rather than a particular object's data.

```python
class student:
    college="GIT college"
    @classmethod       
    def display_college(cls):   #cls--->referece to the class
        print(cls.college)
student.display_college()
```

```python
class animal:
    def eat(self):
        print("Animal is eating")
class dog(animal):
    def bark(self):
        print("Dog is barkig")
d1=dog()
d1.eat()
d1.bark()
```

With Inheritance

```python
class Animal:
    def eat(self):
        print("Eating")
class Dog(Animal):
    def bark(self):
        print("Barking")
class Cat(Animal):
    def meow(self):
        print("meowing")
d1=Dog()
c1=Cat()
d1.eat()
d1.bark()
c1.meow()
```

Types of inheritance
1.single inheritance:one child inherit from one parent class

```python
class Parent:
    def house(self):
        print("This is my house")
class Child(Parent):
    def own(self):
        print("I start a bussiness")
c1=Child()
c1.house()
c1.own()
```

2.Multiple inheritance : one child inherit from more than one parent class

```python
class Father:
    def fp(self):
        print("Father property")
class Mother:
    def mp(self):
        print("Mother property")
class Child(Father,Mother):
    def cm(self):
        print("Child property")
c=Child()
c.fp()
c.mp()
c.cm()
```

Multilevel inheritance : when inheritance happens in multiple levels

```python
class Grandparent:
    def house(self):
        print("Grandparent's house")
class Parent(Grandparent):
    def car(self):
        print("Parent's car")
class Child(Parent):
    def bike(self):
        print("Child's bike")
c=Child()
c.house()
c.car()
c.bike()
    
```

4.Hierarchical inheritance : when multiple child classes inherit from same parent class

```python
class Parent:
    def msg(self):
        print("I'm parent class")
class Child1(Parent):
    def hlo(self):
        print("I'm child1 class")
class Child2(Parent):
    def hlo(self):
        print("I'm child2 class")
class Child3(Parent):
    pass
c1=Child1()
c1.msg()
c1.hlo()
c2=Child2()
c2.msg()
c2.hlo()
```

5.Hybrid inheritance: it is combination of teo or more types of inheritance

```python
class A:
    def msg(self):
        print("I'm A class")
class B(A):
    def msg1(self):
        print("I'm B class")
class C(A):
    def msg2(self):
        print("I'm c class")
class D(B,C):
    def msg3(self):
        print("I'm D class")
d=D()
d.msg()
d.msg1()
d.msg2()
d.msg3()
```
