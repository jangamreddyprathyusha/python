Types of inheritance
1.single inheritance:one child inherit from one parent class

```python
class Parent:
    def house(self):
        print("This is my house")
class Child(Parent):
    def own(self):
        print("I start a bussiness")
c1=Child()
c1.house()
c1.own()
```

2.Multiple inheritance : one child inherit from more than one parent class

```python
class Father:
    def fp(self):
        print("Father property")
class Mother:
    def mp(self):
        print("Mother property")
class Child(Father,Mother):
    def cm(self):
        print("Child property")
c=Child()
c.fp()
c.mp()
c.cm()
```

Multilevel inheritance : when inheritance happens in multiple levels

```python
class Grandparent:
    def house(self):
        print("Grandparent's house")
class Parent(Grandparent):
    def car(self):
        print("Parent's car")
class Child(Parent):
    def bike(self):
        print("Child's bike")
c=Child()
c.house()
c.car()
c.bike()
    
```

4.Hierarchical inheritance : when multiple child classes inherit from same parent class

```python
class Parent:
    def msg(self):
        print("I'm parent class")
class Child1(Parent):
    def hlo(self):
        print("I'm child1 class")
class Child2(Parent):
    def hlo(self):
        print("I'm child2 class")
class Child3(Parent):
    pass
c1=Child1()
c1.msg()
c1.hlo()
c2=Child2()
c2.msg()
c2.hlo()
```

5.Hybrid inheritance: it is combination of two or more types of inheritance

```python
class A:
    def msg(self):
        print("I'm A class")
class B(A):
    def msg1(self):
        print("I'm B class")
class C(A):
    def msg2(self):
        print("I'm c class")
class D(B,C):
    def msg3(self):
        print("I'm D class")
d=D()
d.msg()
d.msg1()
d.msg2()
d.msg3()

```

Abstraction: hide implementation details and show only the necessary functionality.
-->python provides abstraction using the abc module.
-->The child class must provide its own implementation of this method.

```python
from abc import ABC,abstractmethod
class Animal(ABC):  #we can create an abstract class by inheriting from ABC
    @abstractmethod 
    def sound(self):
        pass
```

```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod    #can't create an object if abstract methods are unimplemented.
    def sound(self):
        pass
class Dog(Animal):     #child class must implemented all abstract methods, or remain abstarct.
    def sound(self): 
        print("Dog barks")
class Cat(Animal):
    def sound(self):
        print("Cat meow")
d=Dog()
c=Cat()
d.sound()
c.sound()
```

```python
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
class UPI(Payment):
    def pay(self,amount):
        print("Paid",amount,"using UPI")
class Card(Payment):
    def pay(self,amount):
        print("Paid",amount,"using Card")
u=UPI()
c=Card()
u.pay(4000)
c.pay(500)
```

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Money deposited")
        else:
            print("Invalid amount")
    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Money withdrawn")
        else:
            print("Invalid amount or insufficient balance")
    def get_balance(self):
        return self.__balance
account = BankAccount("John", 5000)
account.deposit(2000)
account.withdraw(1000)
print("Name:", account.name)
print("Balance:", account.get_balance())
```

```python
from abc import ABC,abstractmethod
#abstract class
class Animal(ABC):
    def __init__(self,name):
        self.__name=name
#Encapsulation
    #getter
    def getname(self):
        return self.__name
    #abstract method
    @abstractmethod
    def sound(self):
        pass
    #normal method
    def eat(self):
        print(self.__name,"is eating")
#inheritance
class Dog(Animal):
    #constructor
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed
    #method overriding/polymorphism
    def sound(self):
        print("Dog barks")
class Cat(Animal):
    def sound(self):
        print("Cat meow")
#objects
d=Dog("Tommy","Labrador")
c=Cat("Kitty")
#methods
print(d.getname())
d.sound()
d.eat()

print(c.getname())
c.sound()
c.eat()

        
```

```python
class Number:
    def __init__(self,num):
       self.num=num
    def check_prime(self):
        count=0
        for i in range(1,self.num+1):
            if self.num%i==0:
                 count+=1
        if count==2:
            print(self.num,"is prime")
        else:
            print(self.num,"is not prime")
n=Number(8)
n.check_prime()
```
